```python
import nltk 
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer, ENGLISH_STOP_WORDS
from nltk.util import ngrams
```

```python
corpus = [
    "I love cats",
    "I love dogs",
    "Cats and dogs are great pets"
]
```

# Bag Of Words

```python
vectorizer = CountVectorizer()
```

```python
#Learn Vocabulary and transform documents into feature vectors
bow_matrix = vectorizer.fit_transform(corpus)
```

```python
vocab = vectorizer.get_feature_names_out() 
print("Vocabulary:", vocab)
```

```python
print("\nBOW Vectors:")
print(bow_matrix.toarray())
```

```python
for i , doc_vec in enumerate(bow_matrix.toarray()):
    print(f"Document {i+1} vector: {doc_vec}")
```

# TF-IDF

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus)
```

```python
tfidf_vocab = tfidf_vectorizer.get_feature_names_out() 
print("TF-IDF Vocabulary:", tfidf_vocab)
```

```python
print("\nTF-IDF Vectors:") 
print(tfidf_matrix.toarray())
```

```python
for i , vec in enumerate(tfidf_matrix.toarray()):
    print(f"Document {i+1} TF-IDF vector: {vec}")
```

# N-grams

```python
from nltk import word_tokenize 
sentence = "I love natural language processing"
tokens = word_tokenize(sentence)
print("\nTokens:", tokens)
```

```python
bigrams = list(ngrams(tokens,2))
print("\nBigrams (2-grams):", bigrams)
```

```python
trigrams = list(ngrams(tokens,3))
print("\nTrigrams (3-grams):", trigrams)
```

# Vocabulary Creation

```python
all_text = " ".join(corpus).lower()
#stopwords removal
all_tokens = [word for word in word_tokenize(all_text) if word.isalpha() and word not in ENGLISH_STOP_WORDS]
```

```python
all_tokens
```

```python
#crate sorted unique vocabulary list 
vocabulary = sorted(set(all_tokens))
print("\nVocabulary after stopwords removal:", vocabulary) 
word_to_id = { word:idx for word, idx in enumerate(vocabulary)}
print("\nWord to ID mapping:", word_to_id)
```
